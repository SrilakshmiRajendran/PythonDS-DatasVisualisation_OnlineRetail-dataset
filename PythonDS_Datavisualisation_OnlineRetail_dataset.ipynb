**Task 3 – Feature Engineering**

● Create TotalPrice = Quantity × UnitPrice

● Extract time features (Year, Month, Day, Hour)

● Create categories (Customer Segment, Order Size, Day Type)

```python
Online_Retai_Dataset["TotalPrice"] = (Online_Retai_Dataset["Quantity"] * Online_Retai_Dataset["UnitPrice"])
```

```python
(Online_Retai_Dataset["TotalPrice"]).head()
```

```python
Online_Retai_Dataset.head()
```

```python
#Creation of year,day,month,hour columns
Online_Retai_Dataset["Year"]=Online_Retai_Dataset["InvoiceDate"].dt.year
Online_Retai_Dataset["Year"].head()
```

```python
Online_Retai_Dataset["Month"]=Online_Retai_Dataset["InvoiceDate"].dt.month
```

```python
Online_Retai_Dataset["DayName"]=Online_Retai_Dataset["InvoiceDate"].dt.day_name()
```

```python
Online_Retai_Dataset["DayName"].dtype
```

```python
Online_Retai_Dataset["Hour"]=Online_Retai_Dataset["InvoiceDate"].dt.hour
```

```python
#To see the new created values
Online_Retai_Dataset.tail()
```

```python
Online_Retai_Dataset["TotalPrice"].mean()
```

```python
Online_Retai_Dataset["Country"].unique()
```

```python
Online_Retai_Dataset["CustomerSegment"] = np.where(Online_Retai_Dataset["Country"] == "United Kingdom","Domestic","International")
```

```python
Online_Retai_Dataset["CustomerSegment"].unique()
```

```python
Online_Retai_Dataset.describe()
```

```python
small = Online_Retai_Dataset["Quantity"] < 5

medium = (
    (Online_Retai_Dataset["Quantity"] >= 5) &
    (Online_Retai_Dataset["Quantity"] <= 15)
)

large = Online_Retai_Dataset["Quantity"] > 15
```

```python
print(type(small))
print(type(medium))
print(type(large))
```

```python
Online_Retai_Dataset["OrderSize"] = np.select([small, medium, large],["Small", "Medium", "Large"],default="Unknown")
```

```python
Online_Retai_Dataset["OrderSize"].value_counts()
```

```python
#Ctegory creation using-daytype
Online_Retai_Dataset.describe()
```

```python
Online_Retai_Dataset["DayType"] = np.where(
    Online_Retai_Dataset["DayName"].isin(["Saturday", "Sunday"]),
    "Weekend",
    "Weekday"
)
```

```python
Online_Retai_Dataset["DayType"].value_counts()
```

```python
#line Graph
plt.figure(figsize=(20,5))
plt.bar(Top_Country.index, Top_Country.values)

plt.title("Total Sales by Country")
plt.xlabel("Country")
plt.ylabel("Top Sales")

plt.xticks(rotation=90)
plt.show()
```

```python
#Bar Chart-To plot Best sale month
TopMonth=Online_Retai_Dataset.groupby("Month")["TotalPrice"].sum().sort_values(ascending=False)
TopMonth.head()
```

```python
#Bar chart
plt.figure(figsize=(4,4))
plt.bar(TopMonth.index,TopMonth.values)
plt.title("Top sale in ($) month")
plt.xlabel("Month")
plt.ylabel("Total price")
plt.show()
```

```python
#To check for the highest peak sale
Peaktime_sale=Online_Retai_Dataset.groupby("Hour")["TotalPrice"].sum().sort_values(ascending=False)
Peaktime_sale.head()
```

```python
#Bar plot- to show the peaktime sale
plt.figure(figsize=(4,4))
plt.bar(Peaktime_sale.index,Peaktime_sale.values)
plt.xlabel("Hour")
plt.ylabel("Total Sale")
plt.title("Peak sale in Hour")
plt.show()
```
